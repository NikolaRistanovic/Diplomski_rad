```python
%matplotlib widget
import matplotlib.pyplot as plt
import numpy as np
```

```python
values = np.empty(9,dtype=object)

for i in range(9):
    values[i] = np.load("Measurement_"+str((i+2)*10)+"ma.npy")
```

```python
print(values[0])
```

```python
duty = np.arange(1,99)/100
plt.figure(1)
plt.cla()
for i in range(9):
    plt.plot(duty,values[i],label=str((i+2)*10)+"mA")
plt.plot(duty,12/(1-duty),label="ideal",ls="--")
plt.ylim(10,40)
plt.yticks(np.linspace(10,40,16))
plt.xticks(np.linspace(0,1,21))
plt.title("Vin=12V")
plt.xlabel("Duty cycle")
plt.ylabel("Vout")
plt.axvline(0.5)
plt.legend()
plt.grid()
plt.show()
```
